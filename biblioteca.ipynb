```python
class material:
    def __init__(self,id_material,nombre,tipo):
        self.id_material = id_material
        self.nombre = nombre
        self.tipo = tipo
```

```python
class libro(material):
    def __init__(self,id_material,nombre,tipo,autor,genero,estado):
        super().__init__(id_material,nombre,tipo)
        self.autor=autor
        self.genero=genero
        self.estado=estado

    def registro_libro(self):
        print(f"Libro registrado - id:{self.id_material} - {self.nombre}   autor:{self.autor}   genero:{self.genero}   estado:",end=" ")
        if(self.estado==0):print("Disponible")
        else:print("Prestado")
```

```python
class revista(material):
    def __init__(self,id_material,nombre,tipo,edicion,periodicidad,estado):
        super().__init__(id_material,nombre,tipo)
        self.edicion=edicion
        self.periodicidad=periodicidad
        self.estado=estado
        self.tipo=tipo
    
    def registro_revista(self):
        print(f"Revista registrada - id:{self.id_material} - {self.nombre}   edicion:{self.edicion}   periodo:{self.periodicidad}   estado:",end=" ")
        if(self.estado==0):print("Disponible")
        else:print("Prestado")
```

```python
class m_digital(material):
    def __init__(self,id_material,nombre,tipo,t_archivo,enlace):
        super().__init__(id_material,nombre,tipo)
        self.t_archivo=t_archivo
        self.enlace=enlace

    def registro_m_digital(self):
        print(f"Material Digital registrad0 - id:{self.id_material} - {self.nombre}   tipo de archivo:{self.t_archivo}   enlace:{self.enlace}")
```

```python
class persona:
    def __init__(self,nombre,correo):
        self.nombre=nombre
        self.correo=correo
        
```

```python
class bibliotecario(persona):
    def __init__(self,nombre,correo,No_trabajador,sucursal):
        super().__init__(nombre,correo)
        self.No_trabajador=No_trabajador
        self.sucursal=sucursal

    def agregar_material(self,catalogo,libro,revista,m_digital,opc):
        self.catalago=catalogo
        self.libro=libro
        self.revista=revista
        self.m_digital=m_digital
        self.opc=opc
        
        if self.catalago.id != self.sucursal.No_sucursal:
            print(f"el Bibliotecario:{self.nombre} con ID de trabajador:{self.No_trabajador} en la sucursal:{self.sucursal.No_sucursal}")
            print("No puede agregar un material a un catalogo diferente de su sucursal")
        else:
            print(f"el Bibliotecario:{self.nombre} con ID de trabajador:{self.No_trabajador} en la sucursal:{self.sucursal.No_sucursal}")
            print(f"agrego un nuevo material al catalogo:{self.sucursal.No_sucursal}")
            self.catalago.agregar_catalogo_suc(libro,revista,m_digital,opc)
     
    def confirmar_prestamo(self,prestamo):
        self.prestamo=prestamo

        if self.prestamo.usuario.historial <= 0:
            print(f"usuario {self.prestamo.usuario.nombre} no pudo recivir el material su historial no se lo permite")
        else:
            if self.prestamo.opc == 0: 
                print(f"prestamo autorizado  - usuario {self.prestamo.usuario.nombre} material:{self.prestamo.libro.nombre} fecha de prestamo:{self.prestamo.fecha_prestamo} fecha de entrega:{self.prestamo.fecha_devolucion}")
                self.prestamo.libro.estado = 1;
                self.prestamo.usuario.prestamos.append(self.prestamo.libro)
            if self.prestamo.opc == 1: 
                print(f"prestamo autorizado  - usuario {self.prestamo.usuario.nombre} material:{self.prestamo.revista.nombre} fecha de prestamo:{self.prestamo.fecha_prestamo} fecha de entrega:{self.prestamo.fecha_devolucion}")
                self.prestamo.revista.estado = 1;
                self.prestamo.usuario.prestamos.append(self.prestamo.revista)
            if self.prestamo.opc == 2: 
                print(f"prestamo autorizado  - usuario {self.prestamo.usuario.nombre} material:{self.prestamo.m_digital.nombre} ")
                self.prestamo.m_digital.estado = 1;
                self.prestamo.usuario.prestamos.append(self.prestamo.m_digital)
```

```python
class usuario(persona):
    def __init__(self,nombre,correo,id_usuario,historial):
        super().__init__(nombre,correo)
        self.id_usuario=id_usuario
        self.historial=historial
        self.prestamos = []

    def ver_usuario(self):
        print(f" Se refgistro el usuario:{self.nombre} con ID de usuario:{self.id_usuario} correo:{self.correo}")

    def ver_catalogo(self,catalogo):
        cat=catalogo
        print(f"El usuario:{self.nombre} con ID de usuario: {self.id_usuario} - enlisto los  materiales del catalogo:{cat.id}")
        cat.mostrar_cat(0)

    def mostrar_lista_pretamos(self,nc):
        opc=nc
        print(f"el usuario {self.nombre} tiene los siguientes materiales prestados ")
        for libro in self.prestamos:
            if libro.tipo == 0:print(f"id:{libro.id_material} - {libro.nombre} autor:{libro.autor} genero:{libro.genero}")
        for revista in self.prestamos:
            if revista.tipo==1:print(f"id:{revista.id_material} - {revista.nombre} edicion:{revista.edicion} periodo:{revista.periodicidad}")
        for m_digital in self.prestamos:
            if m_digital.tipo==2:print(f"{m_digital.id_material} {m_digital.nombre}")

    def regresar_material(self,penalizacion,prestamo):
        self.penalizacion = penalizacion
        self.prestamo=prestamo
        ver=self.penalizacion.verificar_fecha(prestamo)
        if(ver==1):
            if(self.prestamo.opc==0):
                print(f"usuario: {self.nombre} regreso el material {prestamo.libro.nombre}", end=" ")
                print(f"devolucion tardia - fecha esperado:{prestamo.fecha_devolucion}  fecha de entrega:{penalizacion.fecha_dev}")
                self.historial-=1
            if(self.prestamo.opc==1):
                print(f"usuario: {self.nombre} regreso el material {prestamo.revista.nombre}", end=" ")
                print(f"devolucion tardia - fecha esperado:{prestamo.fecha_devolucion}  fecha de entrega:{penalizacion.fecha_dev}")
                self.historial-=1
            print(f"tu historial tiene un punto menos - total:{self.historial} ")
        else:
            if(self.prestamo.opc==0):
                print(f"usuario: {self.nombre} regreso el material {prestamo.libro.nombre}", end=" ")
                print(f"devolucion a tiempo - fecha esperado:{prestamo.fecha_devolucion}  fecha de entrega:{penalizacion.fecha_dev}")
                self.historial+=1
            if(self.prestamo.opc==1):
                print(f"usuario: {self.nombre} regreso el material {prestamo.revista.nombre}", end=" ")
                print(f"devolucion a tiempo - fecha esperado:{prestamo.fecha_devolucion}  fecha de entrega:{penalizacion.fecha_dev}")
                self.historial+=1
            print(f"tu historial tiene un punto mas - total:{self.historial} ")
```

```python
class sucursal:

    def __init__(self,No_sucursal,ciudad,catalogo):
        self.No_sucursal=No_sucursal 
        self.ciudad=ciudad
        self.catalogo=catalogo
        

    def mostar_catalago_suc(self):
        opc=self.catalogo.id
        print(f"Sucursal:{self.No_sucursal} con cede en {self.ciudad}")
        self.catalogo.mostrar_cat(opc)
```

```python
class prestamo:
    def __init__(self,usuario,libro,revista,m_digital,opc,fecha_prestamo,fecha_devolucion):
        self.fecha_prestamo=fecha_prestamo
        self.fecha_devolucion=fecha_devolucion
        self.usuario=usuario
        self.libro=libro
        self.revista=revista
        self.m_digital=m_digital
        self.opc=opc

    def mostrar_prestamo(self):
        if self.opc==0:
            print(f"prestamo de el material {self.libro.id_material} nombre:{self.libro.nombre}")
        if self.opc==1:
            print(f"prestamo de el material {self.revista.id_material} nombre:{self.revista.nombre}")
        if self.opc==2:
            print(f"prestamo de el material {self.m_digital.id_material} nombre:{self.m_digital.nombre}")

    
```

```python

class penalizacion:
    def __init__(self,prestamo,fecha_dev):
        self.prestamo=prestamo
        self.fecha_dev = fecha_dev
        #self.usuario = usuario
        #datetime.fecha2=fecha_dev
    
    def verificar_fecha(self,prestamo):
        if prestamo.fecha_devolucion < self.fecha_dev:
            return (1)
        else:
            return(0)
        #print(self.fecha2)
        
            
```

```python
class catalogo:

    def __init__(self,No_catalago,id):
          self.No_catalago = No_catalago
          self.id=id
          self.lista=[]

    def agregar_catalogo_suc(self,libro,revista,m_digital,opc):
            self.opcion=opc
            self.libro=libro
            self.revista=revista
            self.m_digital=m_digital
            if self.opcion == 0:
                self.lista.append(self.libro)
               
            if self.opcion == 1:
                self.lista.append(self.revista)
          
            if self.opcion == 2:
                self.lista.append(self.m_digital)


    def mostrar_cat(self,nc):
        opc=nc
        for libro in self.lista:
            if libro.tipo == 0:
                print(f"id:{libro.id_material} - {libro.nombre} autor:{libro.autor}  genero:{libro.genero}  estado:",end=" ")
                if(libro.estado==0):
                    print("Disponible")
                else:
                    print("Prestado")
        
        for revista in self.lista:
            if revista.tipo==1:
                print(f"id:{revista.id_material} - {revista.nombre} edicion:{revista.edicion} perdiodo:{revista.periodicidad} estado:",end=" ")
                if(revista.estado==0):print("Disponible")
                else:print("Prestado")

        for m_digital in self.lista:
            if m_digital.tipo==2:
                print(f"id:{m_digital.id_material} - {m_digital.nombre}  tipo:{m_digital.t_archivo}  enlace:{m_digital.enlace}")

    def buscar(self,autor,genero,tp,opc):
        self.autor=autor
        self.genero=genero
        self.tp=tp
        self.opc=opc

        for libro in self.lista:
            if self.opc == 0: #si se quiere buscar por autor
                if(libro.tipo==0 and libro.autor==self.autor): print(f" libro encontrado:{libro.id_material} nombre:{libro.nombre} autor:{libro.autor} genero:{libro.genero}")
            if self.opc == 1: #si se quiere  buscar por genero
                if(libro.tipo==0 and libro.genero==self.genero):print(f" libro encontrado:{libro.id_material} nombre:{libro.nombre} autor:{libro.autor} genero:{libro.genero}")
            if self.opc == 2:#si se quiere bucar por tipo
                if(libro.tipo==0 and libro.tipo==self.tp):print(f" libro encontrado:{libro.id_material} nombre:{libro.nombre} autor:{libro.autor} genero:{libro.genero}")
        
        for revista in self.lista:
            if self.opc == 2:#si se quiere bucar por tipo
                if(revista.tipo==1 and revista.tipo==self.tp):print(f" Revista encontrada:{revista.id_material} nombre:{revista.nombre} edicion:{revista.edicion}")

        for m_digital in self.lista:
            if self.opc == 2:#si se quiere bucar por tipo
                if(m_digital.tipo==2 and m_digital.tipo==self.tp):print(f" material digital encontrado:{m_digital.id_material} nombre:{m_digital.nombre} enlace:{m_digital.enlace}")
```
