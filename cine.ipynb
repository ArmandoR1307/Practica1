```python
class persona:
    def __init__(self,nombre,correo):
        self.nombre=nombre
        self.correo=correo
```

```python
class usuario(persona):
    
    usuarios = []

    def __init__(self,nombre,correo,id_usuario):
        super().__init__(nombre,correo)
        self.id_usuario=id_usuario

    def registrar_usuario(self):
            usuario.usuarios.append(self)
            print(f"usuario registrado:_ {self.id_usuario} - Nombre:_ {self.nombre} Correo:_ {self.correo}")

    def reservar_funcion(self,funcion,lugares,promo,promociones):
        self.funcion=funcion
        self.promo = promo
        self.promociones=promociones
        self.lugares=lugares
         
        if (self.funcion.capacidad - lugares) < 0:
              print("No hay la cantidad de asientos disponibles")
        else:
             print(f"El usuario:{self.id_usuario} - {self.nombre} ha reservado {self.lugares} asientos para la funcion: {self.funcion.id_funcion} pelicula:{self.funcion.pelicula.titulo}")
             if(promo == 1):
                  print(f"Promocion activada {promociones.descripcion}")
```

```python
class empleado(persona):
    def __init__(self,nombre,correo,N_trabajador,rol):
        super().__init__(nombre,correo,)
        self.N_trabajador=N_trabajador
        self.rol=rol

    def crear_funcion(self,funcion,id_funcion,pelicula,sala):
        self.funcion=funcion
        self.funcion.capacidad = sala.capacidad
        if self.rol == "administrador":
            self.funcion.id_funcion=id_funcion
            self.funcion.pelicula=pelicula
            self.funcion.sala=sala
            print(f"empleado No:{self.N_trabajador} Rol:{self.rol} - funcion creada - {self.funcion.id_funcion} pelicula:{self.funcion.pelicula.titulo} sala:{self.funcion.sala.No_sala}")
            self.funcion.enlistar_funcion()
        else:
            print(f"empleado No:{self.N_trabajador}  Rol:{self.rol} - no puedes agregar funciones")

    def crear_peli(self,pelicula,titulo,duracion,clasificacion,genero):
        self.pelicula=pelicula
        if self.rol == "administrador":
            self.pelicula.titulo=titulo
            self.pelicula.duracion=duracion
            self.pelicula.clasificacion=clasificacion
            self.pelicula.genero=genero
            print(f"empleado No:{self.N_trabajador} Rol:{self.rol}  - pelicula creada - titulo:{self.pelicula.titulo} duracion:{self.pelicula.duracion} minutos clasificacion:{self.pelicula.clasificacion} genero:{self.pelicula.genero}")
            self.pelicula.enlistar_peli()
        else:
            print(f"empleado No:{self.N_trabajador}  Rol:{self.rol} - no puedes agregar peliculas")

    def crear_promocion(self,promociones,descripcion,condicion):
        self.promociones=promociones
        if self.rol == "administrador":
            self.promociones.descripcion=descripcion
            self.promociones.condicion=condicion
            print(f"empleado No:{self.N_trabajador}  Rol:{self.rol} - promocion agregada - descripcion:{self.promociones.descripcion} condicion:{self.promociones.condicion}")
            self.promociones.enlistar_promo()
        else:
            print(f"empleado No:{self.N_trabajador}  Rol:{self.rol} - no puedes agregar promociones")
```

```python
class Espacio:
    def __init__(self,No_espacio):
        self.No_espacio=No_espacio
        
```

```python
class sala(Espacio):
    def __init__(self,No_espacio,No_sala,tipo,capacidad,disponible):
        super().__init__(No_espacio)
        self.No_sala=No_sala
        self.tipo=tipo
        self.capacidad=capacidad
        self.disponible=disponible
```

```python
class zona_comida(Espacio):
    def __init__(self,No_espacio,nombre):
        super().__init__(No_espacio)
        self.nombre=nombre      
```

```python
class pelicula:

    lista_peli=[]

    def __init__(self,titulo,duracion,clasificacion,genero):
        self.titulo=titulo
        self.duracion=duracion
        self.clasificacion=clasificacion
        self.genero=genero

    def enlistar_peli(self):
        pelicula.lista_peli.append(self)

    @classmethod
    def lista_pelis(cls):
        print(f"Peliculas registradas")
        for pelicula in cls.lista_peli:
            if pelicula.titulo != " ":
                print(f" pelicula:{pelicula.titulo}")
```

```python
class promociones():

    promos=[]

    def __init__(self,descripcion,condicion):
        self.descripcion=descripcion
        self.condicion=condicion

    def enlistar_promo(self):
            promociones.promos.append(self)

    @classmethod
    def promos_reg(cls):
        for promociones in cls.promos:
            if promociones.descripcion != " ":
                print(f"descripcion:{promociones.descripcion}   condiciones:{promociones.condicion}")

```
